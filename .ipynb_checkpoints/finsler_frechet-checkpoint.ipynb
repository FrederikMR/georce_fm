## Loading Values

```python
def load_manifold_data(manifold:str, dim):
    
    save_path = ''.join((path, manifold, '/'))
    
    onlyfiles = [f for f in os.listdir(save_path) if os.path.isfile(os.path.join(save_path, f))]
    onlyfiles = [f for f in onlyfiles if f'batch={batch_size}.pkl' in f]
    
    estimates = {}
    for file in onlyfiles:
        file_path = ''.join((save_path, file))
        with open(file_path, 'rb') as f:
            values = pickle.load(f)
        method = file.split(f'_{manifold}')[0]
        dim = file.replace(f"{method}_", '').replace(manifold, '').replace('.pkl', '').replace('_d=', '').replace(f'_batch={batch_size}', '')
        if dim in estimates:
            estimates[dim] = dict(estimates[dim], **values)
        else:
            estimates[dim] = values
            
    return estimates
    
```

## Functions

```python
def force_fun(z, M):
    
    val = jnp.cos(z)
    
    val2 = jnp.sqrt(jnp.einsum('i,ij,j->', val, M.G(z), val))
    
    return jnp.sin(z)*val/val2
    
v0 = 1.5
```

## Plotting the Frechet Mean

```python
class EllipsoidPlot(ABC):
    def __init__(self, 
                 params=jnp.linspace(0.5,1.0,3),
                ):
        
        self.M = nEllipsoid(dim=2, params=params)
        self.force_fun = lambda z: force_fun(z, self.M)
        self.params = params

    def heat_map(self, X, Y, Z)->Array:
        
        z = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X,Y,Z)
        
        heat_map = vmap(lambda z1: vmap(lambda z: jnp.linalg.norm(self.force_fun(z)))(z1))(z)
        
        return heat_map
    
    def unit_direction(self, n_points:int=100)->Array:
        
        X,Y,Z = self.grid(n_points)
        
        z = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X,Y,Z)
        
        force = vmap(lambda z1: vmap(lambda z: jnp.dot(self.M.Jf(z), self.force_fun(z)))(z1))(z)
        
        return X,Y,Z,force
        
    def grid(self, n_points:int=100)->Array:
        
        theta = jnp.linspace(0.0, jnp.pi*1, n_points)
        phi   = jnp.linspace(0.0, jnp.pi*2, n_points)

        # Creating the coordinate grid for the unit sphere.
        x1 = jnp.outer(jnp.sin(theta), jnp.cos(phi))
        x2 = jnp.outer(jnp.sin(theta), jnp.sin(phi))
        x3 = jnp.outer(jnp.cos(theta), jnp.ones(n_points))
        
        return self.params[0]*x1,self.params[1]*x2,self.params[2]*x3
         
```

```python
class ParaboloidPlot(ABC):
    def __init__(self):
        
        self.M = nParaboloid(dim=2)
        self.force_fun = lambda z: force_fun(z, self.M)
        
        return
    
    def heat_map(self, n_points:int=100)->Array:
        
        X1, X2, X3 = self.grid(n_points)
        X = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X1,X2,X3)
        X1, X2 = X[:,:,0], X[:,:,1]
        
        heat_map = vmap(lambda u1,v1: \
                        vmap(lambda u,v: \
                             jnp.linalg.norm(self.force_fun(jnp.stack((u,v)))))(u1,v1))(X1,X2)
        
        return heat_map
    
    def unit_direction(self, n_points:int=100)->Array:

        X1, X2, X3 = self.grid(n_points)
        X = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X1,X2,X3)
        X1, X2 = X[:,:,0], X[:,:,1]
        
        force = vmap(lambda u1,v1: \
                     vmap(lambda u,v: jnp.dot(self.M.Jf(jnp.stack((u,v))),
                                              self.force_fun(jnp.stack((u,v)))))(u1,v1))(X1,X2)
        
        return X1, X2, X1**2+X2**2, force

    def grid(self, n_points:int=100):
        
        #x1 = jnp.linspace(-2.0, 2.0, n_points)
        #x2 = jnp.linspace(-2.0,2.0, n_points)

        #X1, X2 = jnp.meshgrid(x1,x2)
        
        #return X1, X2, X1**2+X2**2

        u = jnp.linspace(0., 2., n_points)
        v = jnp.linspace(0., 2*jnp.pi, n_points)

        return jnp.outer(u, jnp.cos(v)), jnp.outer(u, jnp.sin(v)), jnp.outer(u**2, jnp.ones_like(u))
        

        
```

```python
class HyperbolicParaboloidPlot(ABC):
    def __init__(self):
        
        self.M = HyperbolicParaboloid()
        self.force_fun = lambda z: force_fun(z, self.M)
        
        return
    
    def heat_map(self, n_points:int=100)->Array:
        
        X1, X2, X3 = self.grid(n_points)
        X = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X1,X2,X3)
        X1, X2 = X[:,:,0], X[:,:,1]
        
        heat_map = vmap(lambda u1,v1: \
                        vmap(lambda u,v: \
                             jnp.linalg.norm(self.force_fun(jnp.stack((u,v)))))(u1,v1))(X1,X2)
        
        return heat_map
    
    def unit_direction(self, n_points:int=100)->Array:
        
        X1, X2, X3 = self.grid(n_points)
        X = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X1,X2,X3)
        X1, X2 = X[:,:,0], X[:,:,1]
        
        force = vmap(lambda u1,v1: \
                     vmap(lambda u,v: jnp.dot(self.M.Jf(jnp.stack((u,v))),
                                              self.force_fun(jnp.stack((u,v)))))(u1,v1))(X1,X2)
        
        return X1, X2, X1**2-X2**2, force

    def grid(self, n_points:int=100):
        
        #x1 = jnp.linspace(-2.0, 2.0, n_points)
        #x2 = jnp.linspace(-2.0,2.0, n_points)
        
        #X1, X2 = jnp.meshgrid(x1,x2)
        
        #return X1, X2, X1**2-X2**2

        u = jnp.linspace(0., 2., n_points)
        v = jnp.linspace(0., 2*jnp.pi, n_points)

        return jnp.outer(u, jnp.cos(v)), jnp.outer(u, jnp.sin(v)), (jnp.outer(u, jnp.cos(v))**2)-(jnp.outer(u, jnp.sin(v))**2)
```

## Conceptual Illustration of Indicatrix Field

```python
iters = 1000
T = 100
tol = 1e-4
rho = 0.5
```

### Generating data and manifold

```python
RM = RS2()
FM = FS2()
```

```python
N_data = 10
```

```python
theta = jnp.linspace(0, 2*jnp.pi, N_data, endpoint=False)
x_obs = 0.1*jnp.vstack((jnp.ones(N_data), jnp.cos(theta), jnp.sin(theta))).T
x_obs = x_obs.T/jnp.linalg.norm(x_obs, axis=1)
x_obs = x_obs.T

z_obs = vmap(RM.invf)(x_obs)
```

```python
xscales = [jnp.min(z_obs[:,0])-0.1, jnp.max(z_obs[:,0])+0.1]
yscales = [jnp.min(z_obs[:,1])-0.1, jnp.max(z_obs[:,1])+0.1]
```

```python
n_points = 5
x1 = jnp.linspace(xscales[0], xscales[1], n_points)
x2 = jnp.linspace(yscales[0], yscales[1], n_points)

X1, X2 = jnp.meshgrid(x1,x2)
```

### Riemannian mean

```python
FrechetMean = RGEORCE_FM(M=RM,
                        init_fun=None,
                        T=T,
                        max_iter=iters,
                        tol = tol,
                        line_search_params={'rho': rho},
                       )
```

```python
fmean_riemannian = FrechetMean(z_obs)
```

```python
xt_rgeodesics = vmap(vmap(RM.f))(fmean_riemannian[1])
xt_rfm = RM.f(fmean_riemannian[0])
rdist = RM.length_frechet(fmean_riemannian[1], z_obs, fmean_riemannian[0])
```

```python
class EllipsoidPlot(ABC):
    def __init__(self, 
                 params=jnp.linspace(0.5,1.0,3),
                ):
        
        self.M = RS2()
        self.force_fun = lambda z: jnp.array([0.75, 0.0])#force_fun(z, self.M)
        self.params = params

    def heat_map(self, X, Y, Z)->Array:
        
        z = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X,Y,Z)
        
        heat_map = vmap(lambda z1: vmap(lambda z: jnp.linalg.norm(self.force_fun(z)))(z1))(z)
        
        return heat_map
    
    def unit_direction(self, n_points:int=100)->Array:
        
        X,Y,Z = self.grid(n_points)
        
        z = vmap(lambda x,y,z: vmap(lambda x1,y1,z1: self.M.invf(jnp.stack((x1,y1,z1))))(x,y,z))(X,Y,Z)
        
        force = vmap(lambda z1: vmap(lambda z: jnp.dot(self.M.Jf(z), self.force_fun(z)))(z1))(z)
        
        return X,Y,Z,force
        
    def grid(self, n_points:int=100)->Array:
        
        theta = jnp.linspace(0.0, jnp.pi*1, n_points)
        phi   = jnp.linspace(0.0, jnp.pi*2, n_points)

        # Creating the coordinate grid for the unit sphere.
        x1 = jnp.outer(jnp.sin(theta), jnp.cos(phi))
        x2 = jnp.outer(jnp.sin(theta), jnp.sin(phi))
        x3 = jnp.outer(jnp.cos(theta), jnp.ones(n_points))
        
        return self.params[0]*x1,self.params[1]*x2,self.params[2]*x3
         
```

```python
class GridPlot(ABC):
    def __init__(self, 
                ):

        return
        
    def grid(self, start_grid=-2.0, end_grid=2.0, n_points=100)->Array:
        
        x1 = jnp.linspace(start_grid, end_grid, n_points)
        x2 = jnp.linspace(start_grid, end_grid, n_points)

        x1, x2 = jnp.meshgrid(x1,x2)
        x3 = jnp.zeros_like(x1)-1.0
        
        return x1, x2, x3
         
```

## Conceptual Illustration of Frechet Mean in the Finlser Case

```python
iters = 1000
T = 100
tol = 1e-4
rho = 0.5
```

### Generating data and manifold

```python
RM = EllipticFinsler(a=lambda x,v: 1, b= lambda x,v: 3, c1= lambda x,v: 0.0, c2= lambda x,v: 0.0, theta= lambda x,v: 0)
FM = EllipticFinsler(a=lambda x,v: 1, b= lambda x,v: 3, c1= lambda x,v: 1/2, c2= lambda x,v: 1/2, theta= lambda x,v: 0)
```

```python
N_data = 10
```

```python
theta = jnp.linspace(0, 2*jnp.pi, N_data, endpoint=False)
z_obs = 0.1*jnp.vstack((jnp.cos(theta), jnp.sin(theta))).T
```

```python
xscales = [jnp.min(z_obs[:,0])-0.1, jnp.max(z_obs[:,0])+0.1]
yscales = [jnp.min(z_obs[:,1])-0.1, jnp.max(z_obs[:,1])+0.1]
```

```python
n_points = 5
x1 = jnp.linspace(xscales[0], xscales[1], n_points)
x2 = jnp.linspace(yscales[0], yscales[1], n_points)

X1, X2 = jnp.meshgrid(x1,x2)
```

### Riemannian mean

```python
FrechetMean = FGEORCE_FM(M=RM,
                        init_fun=None,
                        T=T,
                        max_iter=iters,
                        tol = tol,
                        line_search_params={'rho': rho},
                       )
```

```python
fmean_riemannian = FrechetMean(z_obs)
```

```python
rdist = RM.length_frechet(fmean_riemannian[1], z_obs, fmean_riemannian[0])
```

```python
indicatrix_riemannian = vmap(vmap(lambda u,v: jnp.stack((u,v))+RM.indicatrix(jnp.stack((u,v)), N_points=100)/100.0))(X1,X2) #/50#, grid=jnp.linspace(-5,5,1000), eps=1e-6)
```

### Finsler mean start point

```python
FrechetMean = FGEORCE_FM(M=FM,
                        init_fun=None,
                        T=T,
                        max_iter=iters,
                        tol = tol,
                        line_search_params={'rho': rho},
                       )
```

```python
fmean_finsler = FrechetMean(z_obs)
```

```python
indicatrix_finsler = vmap(vmap(lambda u,v: jnp.stack((u,v))+FM.indicatrix(jnp.stack((u,v)), N_points=100)/100.0))(X1,X2) #/50.0#, grid=jnp.linspace(-5,5,1000), eps=1e-6)
```

### Finsler mean as end point

```python
FrechetMean = FGEORCE_FM_EndPoint(M=FM,
                        init_fun=None,
                        T=T,
                        max_iter=iters,
                        tol = tol,
                        line_search_params={'rho': rho},
                       )
```

```python
fmean_finsler_end_point = FrechetMean(z_obs)
```

```python
indicatrix_finsler = vmap(vmap(lambda u,v: jnp.stack((u,v))+FM.indicatrix(jnp.stack((u,v)), N_points=100)/100.0))(X1,X2) #/50.0#, grid=jnp.linspace(-5,5,1000), eps=1e-6)
```
